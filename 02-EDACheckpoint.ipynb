# Research Question

**Research Question:** How are breed labels and spay/neuter status associated with (1) the likelihood of adoption and (2) the length of stay for animals at the Austin Animal Center?

We will treat adoption as a binary outcome derived from the Outcome Type variable and define length of stay as the time difference between intake and outcome timestamps. We will use statistical inference and predictive modeling to examine how these factors relate to adoption outcomes and length of stay, while also considering other variables such as age and animal type. Because this is observational shelter data, our analysis will focus on identifying associations rather than making causal claims.

## Background and Prior Work

Many animals will unfortunately spend the rest of their lives in animal shelters with their only hope of leaving being adoption. There are numerous reasons that a person may choose to adopt one animal over another but we would like to try to distill these reasons into a few easy to measure variables to determine what factors affect adoption rates most.

A study at UC Davis suggests <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) that one factor that increases how likely an animal is to be adopted is if they are neutered. Which, as a variable that a shelter can control, is worth studying further.

Other studies <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) of participants who adopted animals but later returned them to the shelter concluded that, while dogs are typically returned for aggressive behaviours, cats are most often returned for personal reasons. This may point to the idea that people are more forgiving when it comes to cats having specific characteristics, which may lead to a lower correlation among the measured attributes of a cat and its adoption rate.

This study <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) concluded that size and age are likely the largest contributing factors to how long it is before an animal is adopted. Small dogs, puppies, and large dogs were the fastest group to be adopted, leading to medium sized dogs staying at shelters the longest. Additionally, they found that sex and the color of the animal's coat played a very small role in how likely a dog is to be adopted.

1. <a name="cite_note-1"></a> [^](#cite_ref-1) Jaime Clevenger and Philip H. Kass. (Winter 2003). Determinants of Adoption and Euthanasia of Shelter Dogs Spayed or Neutered in the University of California Veterinary Student Surgery Program Compared to Other Shelter Dogs. 
2. <a name="cite_note-2"></a> [^](#cite_ref-2)  Sloane M Hawes, Josephine M Kerrigan, Tess Hupe, Kevin N Morris. (2020 Sep 3). Factors Informing the Return of Adopted Dogs and Cats to an Animal Shelter. https://pmc.ncbi.nlm.nih.gov/articles/PMC7552273/
3. <a name="cite_note-3"></a> [^](#cite_ref-3) Leslie Sinn. (July–August 2016). Factors affecting the selection of cats by adopters. *Journal of Veterinary Behavior*. https://www.sciencedirect.com/science/article/abs/pii/S1558787816300442
4. <a name="cite_note-4"></a> [^](#cite_ref-4) William P Brown, Janelle P Davidson, Marion E Zuefle. Effects of phenotypic characteristics on the length of stay of dogs at two no kill animal shelters. https://pubmed.ncbi.nlm.nih.gov/23282290/

# Hypothesis


We hypothesize that animals recorded as spayed or neutered will have a higher probability of adoption and a shorter length of stay compared to intact animals. This relationship is expected to vary significantly across breed labels, as historical adoption rates differ between breeds. 

Furthermore, we anticipate that variables such as age and animal type will influence these outcomes and must be controlled in our analysis. We expect these relationships to be associative rather than causal, meaning that while these factors are correlated with adoption speed, they are not the sole drivers of adoption decisions.

```python
import sys
sys.path.append('./modules') # this tells python where to look for modules to import
import pandas as pd
import datetime

df_outcomes = pd.read_csv('https://raw.githubusercontent.com/COGS108/Group072_WI26/refs/heads/master/data/00-raw/Austin%20Animal%20Center%20Outcomes%2010012013%20to%2005052025.csv')
df_intakes = pd.read_csv('https://raw.githubusercontent.com/COGS108/Group072_WI26/refs/heads/master/data/00-raw/Austin%20Animal%20Center%20Intakes%2010012013%20to%2005052025.csv')

#find all the animals which were adopted
df_outcomes = df_outcomes[df_outcomes["outcome_type"] == 'Adoption']

#find all of those which are cats or dogs
df_outcomes = df_outcomes[df_outcomes["animal_type"].isin(['Dog', 'Cat'])]

#remove extra "monthyear" column as well as name, breed, and color
df_outcomes = df_outcomes[['animal_id', 'outcome_type', 'animal_type','sex_upon_outcome', 'age_upon_outcome','datetime', 'date_of_birth']]

#remove duplicate entries
df_outcomes = df_outcomes.drop_duplicates(subset='animal_id', keep=False)

#filter for only dogs and cats
df_intakes = df_intakes[df_intakes["animal_type"].isin(['Dog', 'Cat'])]

#remove unneeded columns
df_intakes = df_intakes[['animal_id', 'datetime', 'intake_type', 'intake_condition','animal_type', 'sex_upon_intake','age_upon_intake']]

#remove duplicate entries 
df_intakes = df_intakes.drop_duplicates(subset='animal_id', keep=False)

#get all of the animal ids which have both intake and outcome entries and drop any duplicate entries
intakes_c = df_intakes[df_intakes['animal_id'].isin(df_outcomes['animal_id'])].drop_duplicates(subset=['animal_id'])
outcomes_c = df_outcomes[df_outcomes['animal_id'].isin(df_intakes['animal_id'])].drop_duplicates(subset=['animal_id'])

#rename datetime columns
intakes_c.columns = ['animal_id', 'intake_datetime', 'intake_type', 'intake_condition','intake_animal_type', 'sex_upon_intake','age_upon_intake']
outcomes_c.columns = ['animal_id', 'outcome_type', 'outcome_animal_type','sex_upon_outcome', 'age_upon_outcome','outcome_datetime', 'date_of_birth']

#properly format dates
intakes_c['intake_datetime'] = pd.to_datetime(intakes_c['intake_datetime'], format='mixed', utc=True)
outcomes_c['outcome_datetime'] = pd.to_datetime(outcomes_c['outcome_datetime'], format='mixed', utc=True)

#combine them together
combined = pd.merge(left=intakes_c, right=outcomes_c, left_on="animal_id", right_on="animal_id")   

#calculate how long the animal spent at the shelter
def calc_length(value):
    return value.total_seconds() / 86400
combined['length_of_stay'] = (combined['outcome_datetime'] - combined['intake_datetime']).apply(calc_length)

#remove bad entries where outcome somehow happens before they are taken in
combined = combined[combined['length_of_stay'] >= 0]
```

#### A deeper exploration into the effect of the animal being fixed

```python
import pandas as pd
import numpy as np
import datetime
import statsmodels
import statsmodels.formula.api as smf
import scipy.stats as stats
from scipy.stats import ttest_ind
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_sort = combined.loc[:, ['sex_upon_outcome','length_of_stay','outcome_animal_type']] #get the columns we want

#convert fixed status to bool
def fixed_status(value):
    if value in ['Neutered Male','Spayed Female']:
        return True
    else:
        return False

df_sort.loc[:, 'is_fixed'] = df_sort['sex_upon_outcome'].apply(fixed_status)
```

Here we need to test for normality to see if we can compare the two populations directly. We will see that the data is not normal, so therefore, we must run a Mann-Whitney U Test.

```python
#test data for normality
cat = df_sort[df_sort['outcome_animal_type'] == 'Cat']
_, cat_p = stats.normaltest(cat[cat['is_fixed'] == False]['length_of_stay'])
print(f"the p-value for cats being normal: {cat_p}")

dog = df_sort[df_sort['outcome_animal_type'] == 'Dog']
_, dog_p =stats.normaltest(dog[dog['is_fixed'] == True]['length_of_stay'])
print(f"the p-value for dogs data being normal: {dog_p}")
```

```python
#since data is not normal we use Mann‑Whitney U test
fixed = df_sort[df_sort['is_fixed'] == True]
not_fixed = df_sort[df_sort['is_fixed'] == False]

fixed_dog = fixed[fixed['outcome_animal_type'] == 'Dog']
not_fixed_dog = not_fixed[not_fixed['outcome_animal_type'] == 'Dog']

fixed_cat = fixed[fixed['outcome_animal_type'] == 'Cat']
not_fixed_cat = not_fixed[not_fixed['outcome_animal_type'] == 'Cat']

_, mwu_dog = stats.mannwhitneyu(fixed_dog['length_of_stay'], not_fixed_dog['length_of_stay'])
_, mwu_cat = stats.mannwhitneyu(fixed_cat['length_of_stay'], not_fixed_cat['length_of_stay'])

print(f"the p-value for fixed status's effect on length of stay for dogs: {mwu_dog}")
print(f"the p-value for fixed status's effect on length of stay for cats: {mwu_cat}")
#seems that fixed status does not have a statistically significant effect on dogs length of stay but it does for cats
```

We can see that the fixed status's influence on the length of stay for dogs can be explained by random chance, however the fixed status of a cat does definitely have an effect on how long it stays. To see this further, lets look at bar graphs of the two.

```python
dog_bar_data = dog[['is_fixed','length_of_stay']].groupby('is_fixed').agg('mean')
sns.barplot(data=dog_bar_data['length_of_stay'])
plt.title('Dogs: Length of Stay vs Fixed Status')
plt.xlabel('is the dog fixed')
plt.ylabel('length of stay in shelter (days)')
plt.show()
```

We can see here that there is a slight increase in the length of stay in days for dogs that are fixed compared to ones that are not. However, since our p-value was 0.2, we see that this difference could easily be explained by random chance.

```python
cat_bar_data = cat[['is_fixed','length_of_stay']].groupby('is_fixed').agg('mean')
sns.barplot(data=cat_bar_data['length_of_stay'])
plt.title('Cats: Length of Stay vs Fixed Status')
plt.xlabel('is the cat fixed')
plt.ylabel('length of stay in shelter (days)')
plt.show()
```

Looking at the bar graph for the lengh of stay vs fixed status for cats, we can see a much bigger difference here. Our p-value for this test was unbelievably small, so we know that this difference is populations is not due to random chance, and instead is caused by something.

```python
#since this is a no kill shelter, there are lots of large outliers for length of stay so to make this chart readable they are not displayed
sns.boxplot(data=df_sort, x='outcome_animal_type', y='length_of_stay', hue='is_fixed', showfliers=False)
plt.title('Length of Stay vs Fixed Status')
plt.xlabel('animal type')
plt.ylabel('length of stay in shelter (days)')
plt.show()
```

Looking at the four populations (Dog Fixed, Dog Unfixed, Cat Fixed and Cat Unfixed) in a boxplot, we can see how different the effect for cats compared to dogs. We can see that there's little effect on the population means for dogs, however, the distribution seems to be different, with a much wider range for dogs that are neutered compared to ones that are not.

This could once again possibly be explained by the fact that older dogs are more likely to be neutered, and this could just be a confounding variable for age.
