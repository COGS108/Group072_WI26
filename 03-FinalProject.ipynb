## Research Question

**Research Question:** How are spay/neuter status, gender, age, and animal type associated with the length of stay for animals at the Austin Animal Center?

We will define length of stay as the time difference between intake and outcome timestamps. We will use statistical inference and predictive modeling to examine how these factors relate to the length of the animal's stay, while also considering other variables such as age and animal type. Because this is observational shelter data, our analysis will focus on identifying associations rather than making causal claims.

## Background and Prior Work

Many animals will unfortunately spend the rest of their lives in animal shelters with their only hope of leaving being adoption. There are numerous reasons that a person may choose to adopt one animal over another but we would like to try to distill these reasons into a few easy to measure variables to determine what factors affect adoption rates most.

A study at UC Davis suggests <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) that one factor that increases how likely an animal is to be adopted is if they are neutered. Which, as a variable that a shelter can control, is worth studying further.

Other studies <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) of participants who adopted animals but later returned them to the shelter concluded that, while dogs are typically returned for aggressive behaviours, cats are most often returned for personal reasons. This may point to the idea that people are more forgiving when it comes to cats having specific characteristics, which may lead to a lower correlation among the measured attributes of a cat and its adoption rate.

This study <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) concluded that size and age are likely the largest contributing factors to how long it is before an animal is adopted. Small dogs, puppies, and large dogs were the fastest group to be adopted, leading to medium sized dogs staying at shelters the longest. Additionally, they found that sex and the color of the animal's coat played a very small role in how likely a dog is to be adopted.

## Hypothesis


We hypothesize that animals recorded as spayed or neutered will have a higher probability of adoption and therefore a shorter length of stay compared to intact animals. This relationship is expected to vary significantly across breed labels, as historical adoption rates differ between breeds. 

Furthermore, we anticipate that variables such as age and animal type will influence these outcomes and must be controlled in our analysis. We expect these relationships to be associative rather than causal, meaning that while these factors are correlated with adoption speed, they are not the sole drivers of adoption decisions.

## Data

### Dataset Overview

We use the **Austin Animal Center Intakes and Outcomes** data, which records animals entering and leaving the Austin Animal Center shelter system.

The raw data comes from two related files stored in our project repository:

- `Austin Animal Center Intakes 10012013 to 05052025.csv`
- `Austin Animal Center Outcomes 10012013 to 05052025.csv`

Each observation corresponds to an individual shelter record. The two files are linked by a shared identifier, `animal_id`, which allows intake and outcome records for the same animal to be matched.

For this project, we focus on **dogs and cats with adoption outcomes**, since our research question asks which factors are associated with an animal's **length of stay before adoption**.


### Key Variables

Important raw variables from the source data include:

- `animal_id`: unique identifier used to merge intake and outcome records
- `outcome_type`: final shelter outcome; used to restrict the data to adoptions
- `datetime`: timestamp field in the raw files, later separated into intake and outcome datetimes
- `date_of_birth`: animal birth date recorded in the outcomes file
- `animal_type`: animal category, such as Dog or Cat
- `sex_upon_intake` / `sex_upon_outcome`: sex labels that also contain spay/neuter information
- `age_upon_intake` / `age_upon_outcome`: age values stored as strings, used to control for age effects
- `intake_type` and `intake_condition`: shelter intake context variables

After cleaning and merging, the main analysis variables are:

- `intake_datetime`
- `outcome_datetime`
- `length_of_stay`
- `outcome_animal_type`
- `sex_upon_outcome`
- parsed `gender`
- derived `is_fixed`
- parsed age in days and age categories

### Data Wrangling

The raw shelter data required several preprocessing steps before it could be used for analysis.

First, we filtered the outcomes table to keep only **adoptions** and restricted both intake and outcome records to **dogs and cats**. We then removed unnecessary columns and matched intake and outcome records using `animal_id`.

Next, we converted date columns to datetime format and computed a derived variable, `length_of_stay`, as the time difference between intake and outcome. We removed invalid records with negative stay durations.

We also created analysis-specific variables from the cleaned data:
- **gender**, parsed from `sex_upon_outcome`
- **fixed status**, based on labels such as `"Neutered Male"` and `"Spayed Female"`
- **age in days**, parsed from string values such as `"2 years"` or `"3 weeks"`
- **age categories** to make comparisons easier across groups

This wrangling was necessary because the raw data was split across two files, contained repeated shelter records, and stored age and status information in semi-structured text fields.

### Cleaning Challenges and Edge Cases

This dataset involved several non-trivial wrangling challenges:

- Intake and outcome information were stored in separate files and had to be merged carefully.
- Some animals appeared more than once because of repeat shelter visits.
- Age values were stored as strings rather than as numeric values, so they required custom parsing.
- Sex labels mixed biological sex with spay/neuter status, requiring additional interpretation.
- A small number of records produced impossible values, such as negative length of stay, and were removed.

Because the project relies on derived variables rather than a single ready-to-use table, careful cleaning was an important part of the workflow.

### 2. Gender Analysis

Next, we examine whether an animal’s gender is associated with its length of stay. Since gender is encoded within the `sex_upon_outcome` field, we first extract a simplified gender variable and remove unknown values.

```python
# Create gender dataframe
df_gender = df.loc[:, ['sex_upon_outcome', 'length_of_stay', 'outcome_animal_type']]

# Extract gender
def get_gender(value):
    if 'Male' in str(value):
        return 'Male'
    elif 'Female' in str(value):
        return 'Female'
    else:
        return 'Unknown'

df_gender['gender'] = df_gender['sex_upon_outcome'].apply(get_gender)

# Remove unknown values
df_gender = df_gender[df_gender['gender'] != 'Unknown']

# Split by animal type
cat_data = df_gender[df_gender['outcome_animal_type'] == 'Cat']
dog_data = df_gender[df_gender['outcome_animal_type'] == 'Dog']
```

We compare the average length of stay between male and female animals using bar plots.

```python
dog_bar_data = dog_data[['gender', 'length_of_stay']].groupby('gender').agg('mean').reset_index()

sns.barplot(data=dog_bar_data, x='gender', y='length_of_stay', hue='gender', palette='Blues_d', legend=False)
plt.title('Dogs: Average Length of Stay vs Gender')
plt.xlabel('Gender')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For dogs, the average length of stay appears similar across genders, suggesting that gender may not be a strong factor influencing adoption timing.

```python
cat_bar_data = cat_data[['gender', 'length_of_stay']].groupby('gender').agg('mean').reset_index()

sns.barplot(data=cat_bar_data, x='gender', y='length_of_stay', hue='gender', palette='Oranges_d', legend=False)
plt.title('Cats: Average Length of Stay vs Gender')
plt.xlabel('Gender')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For cats, a similar pattern emerges, with little difference in average length of stay between male and female animals. This suggests that gender is not a major factor in determining how quickly cats are adopted.

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df_gender, x='outcome_animal_type', y='length_of_stay', hue='gender')

plt.title('Length of Stay by Animal Type and Gender')
plt.xlabel('Animal Type')
plt.ylabel('Length of Stay (Days)')
plt.legend(title='Gender')
plt.show()
```

Across both animal types, gender does not appear to strongly influence length of stay. This suggests that adopters are likely prioritizing other characteristics, such as age or health status, over gender when choosing an animal.

```python
male_dogs = dog_data[dog_data['gender'] == 'Male']['length_of_stay']
female_dogs = dog_data[dog_data['gender'] == 'Female']['length_of_stay']

t_stat, p_val = stats.ttest_ind(male_dogs, female_dogs, equal_var=False)

print("Dogs t-test p-value:", p_val)
```

```python
male_cats = cat_data[cat_data['gender'] == 'Male']['length_of_stay']
female_cats = cat_data[cat_data['gender'] == 'Female']['length_of_stay']

t_stat, p_val = stats.ttest_ind(male_cats, female_cats, equal_var=False)

print("Cats t-test p-value:", p_val)
```

To further evaluate whether gender is associated with length of stay, we perform a two-sample t-test comparing male and female animals. 

The resulting p-values are extremely small for both dogs and cats, indicating that we reject the null hypothesis and conclude that gender is statistically associated with differences in length of stay.

However, the differences observed in the bar plots are relatively small in magnitude. This suggests that while the effect is statistically significant, it may not be practically meaningful. In other words, gender appears to have a measurable but limited impact on adoption timing. This may be driven by the large sample size, which allows even small differences to appear statistically significant.

### 3. Spay/Neuter Status Analysis

We next examine whether spay/neuter status is associated with an animal’s length of stay. Since this information is encoded within the `sex_upon_outcome` variable, we first extract a binary indicator for whether an animal is fixed.

```python
# Create dataframe
df_sort = df.loc[:, ['sex_upon_outcome', 'length_of_stay', 'outcome_animal_type']]

# Convert to fixed status
def fixed_status(value):
    return value in ['Neutered Male', 'Spayed Female']

df_sort['is_fixed'] = df_sort['sex_upon_outcome'].apply(fixed_status)

# Split groups
fixed = df_sort[df_sort['is_fixed']]
not_fixed = df_sort[~df_sort['is_fixed']]

fixed_dog = fixed[fixed['outcome_animal_type'] == 'Dog']
not_fixed_dog = not_fixed[not_fixed['outcome_animal_type'] == 'Dog']

fixed_cat = fixed[fixed['outcome_animal_type'] == 'Cat']
not_fixed_cat = not_fixed[not_fixed['outcome_animal_type'] == 'Cat']
```

We first test whether the data follows a normal distribution. The results indicate strong deviations from normality (p ≈ 0), so we use a non-parametric Mann–Whitney U test to compare groups.

```python
_, mwu_dog = stats.mannwhitneyu(
    fixed_dog['length_of_stay'],
    not_fixed_dog['length_of_stay']
)

_, mwu_cat = stats.mannwhitneyu(
    fixed_cat['length_of_stay'],
    not_fixed_cat['length_of_stay']
)

print("Dogs p-value:", mwu_dog)
print("Cats p-value:", mwu_cat)
```

For dogs, the p-value is relatively large, indicating that we fail to reject the null hypothesis. This suggests that spay/neuter status does not have a statistically significant effect on length of stay for dogs.

For cats, the p-value is extremely small, indicating a statistically significant difference in length of stay between fixed and non-fixed animals.

```python
dog_bar_data = df_sort[df_sort['outcome_animal_type'] == 'Dog'] \
    .groupby('is_fixed')['length_of_stay'].mean().reset_index()

sns.barplot(data=dog_bar_data, x='is_fixed', y='length_of_stay')
plt.title('Dogs: Length of Stay vs Fixed Status')
plt.xlabel('Is Fixed')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For dogs, the difference in average length of stay between fixed and non-fixed animals is relatively small, which aligns with the non-significant statistical result.

```python
cat_bar_data = df_sort[df_sort['outcome_animal_type'] == 'Cat'] \
    .groupby('is_fixed')['length_of_stay'].mean().reset_index()

sns.barplot(data=cat_bar_data, x='is_fixed', y='length_of_stay')
plt.title('Cats: Length of Stay vs Fixed Status')
plt.xlabel('Is Fixed')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For cats, there is a much larger difference in average length of stay between fixed and non-fixed animals, consistent with the statistically significant result.

```python
sns.boxplot(
    data=df_sort,
    x='outcome_animal_type',
    y='length_of_stay',
    hue='is_fixed',
    showfliers=False
)

plt.title('Length of Stay by Animal Type and Fixed Status')
plt.xlabel('Animal Type')
plt.ylabel('Length of Stay (Days)')
plt.show()
```
