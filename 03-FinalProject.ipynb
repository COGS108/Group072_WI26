## Abstract

Animal shelters aim to minimize the amount of time animals spend waiting to be adopted, but adoption timelines can vary widely depending on an animal’s characteristics. In this project, we investigate how spay/neuter status, gender, age, and animal type are associated with the length of stay for animals at the Austin Animal Center. Using publicly available shelter data, we focus on cats and dogs with recorded adoption outcomes and define length of stay as the time between intake and adoption. We apply exploratory data analysis, statistical tests, and regression models to examine how these factors relate to differences in adoption timing.

Our results show that age is one of the strongest factors associated with length of stay, particularly for dogs, where middle-aged animals tend to remain in shelters longer than younger or older ones. For cats, age also plays a role, though the relationship is weaker. Spay/neuter status does not have a statistically significant effect for dogs, but is associated with longer stays for cats, potentially reflecting confounding factors such as age or behavioral history. Overall, these findings highlight how certain observable characteristics are linked to adoption timelines and suggest that shelters may be able to use this information to better support animals that are at higher risk of extended stays.

## Research Question

**Research Question:** How are spay/neuter status, gender, age, and animal type associated with the length of stay for cats and dogs at the Austin Animal Center?

We will define length of stay as the time difference between intake and outcome timestamps. Although the dataset contains multiple animal types, this analysis focuses specifically on cats and dogs due to their prevalence and relevance to adoption patterns. We use statistical tests and regression analysis to examine how these factors relate to length of stay. Because this is observational shelter data, our analysis focuses on identifying associations rather than making causal claims.

## Background and Prior Work

A study at UC Davis suggests <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) that one factor that increases how likely an animal is to be adopted is if they are neutered. Which, as a variable that a shelter can control, is worth studying further.

## Hypothesis


We hypothesize that animals recorded as spayed or neutered will tend to have shorter lengths of stay compared to intact animals. We further expect that age and animal type may be associated with length of stay, with older animals potentially remaining in shelters longer. ## Data

### Dataset Overview

We use the **Austin Animal Center Intakes and Outcomes** data, which records animals entering and leaving the Austin Animal Center shelter system.

The raw data comes from two related files stored in our project repository:

- `Austin Animal Center Intakes 10012013 to 05052025.csv`
- `Austin Animal Center Outcomes 10012013 to 05052025.csv`

Each observation corresponds to an individual shelter record. The two files are linked by a shared identifier, `animal_id`, which allows intake and outcome records for the same animal to be matched.

For this project, we focus on **cats and dogs with adoption outcomes**, since our research question asks which factors are associated with an animal's **length of stay prior to adoption**. The cleaned dataset contains approximately 64,829 observations after filtering for these animals.


### Key Variables

Important raw variables from the source data include:

- `animal_id`: unique identifier used to merge intake and outcome records
- `outcome_type`: final shelter outcome; used to restrict the data to adoptions
- `datetime`: timestamp field in the raw files, later separated into intake and outcome datetimes
- `date_of_birth`: animal birth date recorded in the outcomes file
- `animal_type`: animal category, such as Dog or Cat
- `sex_upon_intake` / `sex_upon_outcome`: sex labels that also contain spay/neuter information
- `age_upon_intake` / `age_upon_outcome`: age values stored as strings, used to control for age effects
- `intake_type` and `intake_condition`: shelter intake context variables

After cleaning and merging, the main analysis variables are:

- `intake_datetime`
- `outcome_datetime`
- `length_of_stay`
- `outcome_animal_type`
- `sex_upon_outcome`
- parsed `gender`
- derived `is_fixed`
- parsed age in days and age categories

### Data Wrangling

The raw shelter data required several preprocessing steps before it could be used for analysis.

First, we filtered the outcomes table to keep only **adoptions** and restricted both intake and outcome records to **dogs and cats**. We then removed unnecessary columns and matched intake and outcome records using `animal_id`.

Next, we converted date columns to datetime format and computed a derived variable, `length_of_stay`, as the time difference between intake and outcome. We removed invalid records with negative stay durations.

We also created analysis-specific variables from the cleaned data:
- **gender**, parsed from `sex_upon_outcome`
- **fixed status**, based on labels such as `"Neutered Male"` and `"Spayed Female"`
- **age in days**, parsed from string values such as `"2 years"` or `"3 weeks"`
- **age categories** to make comparisons easier across groups

This wrangling was necessary because the raw data was split across two files, contained repeated shelter records, and stored age and status information in semi-structured text fields.

### Cleaning Challenges and Edge Cases

This dataset involved several non-trivial wrangling challenges:

- Intake and outcome information were stored in separate files and had to be merged carefully.
- Some animals appeared more than once because of repeat shelter visits.
- Age values were stored as strings rather than as numeric values, so they required custom parsing.
- Sex labels mixed biological sex with spay/neuter status, requiring additional interpretation.
- A small number of records produced impossible values, such as negative length of stay, and were removed.

Because the project relies on derived variables rather than a single ready-to-use table, careful cleaning was an important part of the workflow.

### 2. Gender Analysis

Next, we examine whether an animal’s gender is associated with its length of stay. Since gender is encoded within the `sex_upon_outcome` field, we first extract a simplified gender variable and remove unknown values.

```python
# Create gender dataframe
df_gender = df.loc[:, ['sex_upon_outcome', 'length_of_stay', 'outcome_animal_type']]

# Extract gender
def get_gender(value):
    if 'Male' in str(value):
        return 'Male'
    elif 'Female' in str(value):
        return 'Female'
    else:
        return 'Unknown'

df_gender['gender'] = df_gender['sex_upon_outcome'].apply(get_gender)

# Remove unknown values
df_gender = df_gender[df_gender['gender'] != 'Unknown']

# Split by animal type
cat_data = df_gender[df_gender['outcome_animal_type'] == 'Cat']
dog_data = df_gender[df_gender['outcome_animal_type'] == 'Dog']
```

We compare the average length of stay between male and female animals using bar plots.

```python
dog_bar_data = dog_data[['gender', 'length_of_stay']].groupby('gender').agg('mean').reset_index()

sns.barplot(data=dog_bar_data, x='gender', y='length_of_stay', hue='gender', palette='Blues_d', legend=False)
plt.title('Dogs: Average Length of Stay vs Gender')
plt.xlabel('Gender')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For dogs, the average length of stay appears similar across genders, suggesting that gender may not be a strong factor influencing adoption timing.

```python
cat_bar_data = cat_data[['gender', 'length_of_stay']].groupby('gender').agg('mean').reset_index()

sns.barplot(data=cat_bar_data, x='gender', y='length_of_stay', hue='gender', palette='Oranges_d', legend=False)
plt.title('Cats: Average Length of Stay vs Gender')
plt.xlabel('Gender')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For cats, a similar pattern emerges, with little difference in average length of stay between male and female animals. This suggests that gender is not a major factor in determining how quickly cats are adopted.

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df_gender, x='outcome_animal_type', y='length_of_stay', hue='gender')

plt.title('Length of Stay by Animal Type and Gender')
plt.xlabel('Animal Type')
plt.ylabel('Length of Stay (Days)')
plt.legend(title='Gender')
plt.show()
```

Across both animal types, gender does not appear to strongly influence length of stay. This suggests that adopters are likely prioritizing other characteristics, such as age or health status, over gender when choosing an animal.

```python
male_dogs = dog_data[dog_data['gender'] == 'Male']['length_of_stay']
female_dogs = dog_data[dog_data['gender'] == 'Female']['length_of_stay']

t_stat, p_val = stats.ttest_ind(male_dogs, female_dogs, equal_var=False)

print("Dogs t-test p-value:", p_val)
```

```python
male_cats = cat_data[cat_data['gender'] == 'Male']['length_of_stay']
female_cats = cat_data[cat_data['gender'] == 'Female']['length_of_stay']

t_stat, p_val = stats.ttest_ind(male_cats, female_cats, equal_var=False)

print("Cats t-test p-value:", p_val)
```

To further evaluate whether gender is associated with length of stay, we perform a two-sample t-test comparing male and female animals. 

The resulting p-values are extremely small for both dogs and cats, indicating that we reject the null hypothesis and conclude that gender is statistically associated with differences in length of stay.

However, the differences observed in the bar plots are relatively small in magnitude. This suggests that while the effect is statistically significant, it may not be practically meaningful. In other words, gender appears to have a measurable but limited impact on adoption timing. This may be driven by the large sample size, which allows even small differences to appear statistically significant.

### 3. Spay/Neuter Status Analysis

We next examine whether spay/neuter status is associated with an animal’s length of stay. Since this information is encoded within the `sex_upon_outcome` variable, we first extract a binary indicator for whether an animal is fixed.

```python
# Create dataframe
df_sort = df.loc[:, ['sex_upon_outcome', 'length_of_stay', 'outcome_animal_type']]

# Convert to fixed status
def fixed_status(value):
    return value in ['Neutered Male', 'Spayed Female']

df_sort['is_fixed'] = df_sort['sex_upon_outcome'].apply(fixed_status)

# Split groups
fixed = df_sort[df_sort['is_fixed']]
not_fixed = df_sort[~df_sort['is_fixed']]

fixed_dog = fixed[fixed['outcome_animal_type'] == 'Dog']
not_fixed_dog = not_fixed[not_fixed['outcome_animal_type'] == 'Dog']

fixed_cat = fixed[fixed['outcome_animal_type'] == 'Cat']
not_fixed_cat = not_fixed[not_fixed['outcome_animal_type'] == 'Cat']
```

We first test whether the data follows a normal distribution. The results indicate strong deviations from normality (p ≈ 0), so we use a non-parametric Mann–Whitney U test to compare groups.

```python
_, mwu_dog = stats.mannwhitneyu(
    fixed_dog['length_of_stay'],
    not_fixed_dog['length_of_stay']
)

_, mwu_cat = stats.mannwhitneyu(
    fixed_cat['length_of_stay'],
    not_fixed_cat['length_of_stay']
)

print("Dogs p-value:", mwu_dog)
print("Cats p-value:", mwu_cat)
```

For dogs, the p-value is relatively large, indicating that we fail to reject the null hypothesis. This suggests that spay/neuter status does not have a statistically significant effect on length of stay for dogs.

For cats, the p-value is extremely small, indicating a statistically significant difference in length of stay between fixed and non-fixed animals.

```python
dog_bar_data = df_sort[df_sort['outcome_animal_type'] == 'Dog'] \
    .groupby('is_fixed')['length_of_stay'].mean().reset_index()

sns.barplot(data=dog_bar_data, x='is_fixed', y='length_of_stay')
plt.title('Dogs: Length of Stay vs Fixed Status')
plt.xlabel('Is Fixed')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For dogs, the difference in average length of stay between fixed and non-fixed animals is relatively small, which aligns with the non-significant statistical result.

```python
cat_bar_data = df_sort[df_sort['outcome_animal_type'] == 'Cat'] \
    .groupby('is_fixed')['length_of_stay'].mean().reset_index()

sns.barplot(data=cat_bar_data, x='is_fixed', y='length_of_stay')
plt.title('Cats: Length of Stay vs Fixed Status')
plt.xlabel('Is Fixed')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

For cats, there is a much larger difference in average length of stay between fixed and non-fixed animals, consistent with the statistically significant result.

```python
sns.boxplot(
    data=df_sort,
    x='outcome_animal_type',
    y='length_of_stay',
    hue='is_fixed',
    showfliers=False
)

plt.title('Length of Stay by Animal Type and Fixed Status')
plt.xlabel('Animal Type')
plt.ylabel('Length of Stay (Days)')
plt.show()
```

To improve visualization clarity, extreme outliers in length of stay are not displayed in the boxplot. However, these values are still included in all statistical analyses.

The boxplot highlights differences in distribution across groups. For dogs, the central tendencies are similar, though the spread differs slightly. For cats, fixed animals tend to have noticeably longer stays.

However, this effect may be influenced by **confounding with age**, since older animals are more likely to be neutered and also tend to remain in shelters longer. This suggests that the observed relationship may not be purely causal.

### 4. Age Analysis

Finally, we examine whether age is associated with length of stay. Because age is stored as text values such as `"2 years"` or `"3 months"`, we first convert it into numeric days and then group animals into broader age categories.

```python
def parse_age_to_days(age_str):
    if pd.isna(age_str):
        return np.nan
    parts = str(age_str).split()
    if len(parts) < 2:
        return np.nan
    try:
        val = int(parts[0])
        unit = parts[1].lower()
        if 'year' in unit:
            return val * 365
        elif 'month' in unit:
            return val * 30
        elif 'week' in unit:
            return val * 7
        elif 'day' in unit:
            return val
    except:
        return np.nan
    return np.nan


def age_to_category(age):
    if age <= 365:
        return "0-1"
    elif age <= 3 * 365:
        return "1-3"
    elif age <= 7 * 365:
        return "3-7"
    elif age <= 15 * 365:
        return "7-15"
    else:
        return "15+"


age_df = df.loc[:, ['age_upon_intake', 'length_of_stay', 'outcome_animal_type']].copy()

age_df['age'] = age_df['age_upon_intake'].apply(parse_age_to_days)
age_df = age_df[age_df['age'] >= 0]
age_df['age_category'] = age_df['age'].apply(age_to_category)
```

To improve readability, extreme outliers are not displayed in the boxplot below, though they remain included in the statistical models.

```python
sns.boxplot(
    data=age_df,
    x='outcome_animal_type',
    y='length_of_stay',
    hue='age_category',
    showfliers=False
)

plt.title('Length of Stay vs Animal Age')
plt.xlabel('Animal Type')
plt.ylabel('Length of Stay (Days)')
plt.legend(title='Age Category (Years)')
plt.show()
```

For dogs, age appears to have a strong relationship with length of stay. In general, older dogs tend to remain in the shelter longer before adoption. One interesting exception is the `15+` group, which appears to have shorter stays than the `7-15` group.

For cats, age also appears to matter, but the pattern is less visually clear than it is for dogs.

To evaluate these patterns more formally, we fit separate OLS models for dogs and cats using age category as a predictor of length of stay.

```python
dog_model = smf.ols(
    formula="length_of_stay ~ C(age_category)",
    data=age_df[age_df["outcome_animal_type"] == "Dog"]
).fit()

dog_model.summary()
```

For dogs, the overall model is highly significant, indicating that age category is associated with length of stay. Relative to the youngest group (`0-1`), the `1-3`, `3-7`, and `7-15` groups all show longer stays, with the `7-15` group showing the largest increase. The `15+` group is not statistically distinguishable from the youngest group in this model, which matches the visual pattern in the boxplot.

```python
cat_model = smf.ols(
    formula="length_of_stay ~ C(age_category)",
    data=age_df[age_df["outcome_animal_type"] == "Cat"]
).fit()

cat_model.summary()
```

For cats, the overall model is also statistically significant, though the explanatory power is weaker than for dogs. All older age categories show longer stays than the `0-1` group, and the largest increase appears in the `15+` group. This suggests that age still matters for cats, but the relationship is less pronounced than it is for dogs.

Taken together, these results suggest that age is one of the strongest predictors of length of stay in this project, especially for dogs. This also helps explain earlier findings in the correlation and fixed-status sections, where age likely acted as an important confounding variable.

## Discussion and Conclusion

If only we could adopt every animal, so that no more would be stuck in shelters. Unfortunately, this is impossible; however, throughout this project we have looked deeper into what factors are associated with some animals being adopted quickly compared to others that spend years waiting to find their forever home.

The first thing that stood out the most was that age made a stronger difference for dog adoptions than for cat adoptions. Young dogs tend to have shorter lengths of stay than middle-aged dogs. Older dogs find owners quite quickly as well, but not as quickly as puppies. This may be due to the fact that puppies are incredibly cute and playful, and will be around the family for a much longer period of time. Older dogs may be adopted more quickly for a different reason: they are seen as less of a commitment, and are often more mellow, which can appeal to newer pet owners who might be worried about a more energetic younger dog. Middle-aged dogs have fewer of the advantages of either younger or older dogs, which could explain their longer stays.

Cats, on the other hand, show a weaker relationship between age and length of stay. While age still appears to matter, the differences between age groups are less pronounced than they are for dogs.

Surprisingly, for both cats and dogs, animals that were fixed tended to have longer lengths of stay than those that were not. This makes some sense for dogs, as fixed status is associated with age (older animals are more likely to be fixed). Because younger dogs tend to be adopted more quickly, they are also less likely to be fixed, which may explain this result.

The pattern for cats is more surprising. Age appears to be a weaker factor for cats, so it may not fully explain this relationship. One possible explanation could be behavioral issues. All animals (including humans) are influenced by hormones, and these can affect behavior. If a cat is frequently misbehaving, one possible response is to have it spayed or neutered. However, this may not resolve the underlying issue. It is therefore possible that fixed cats are more likely to have behavioral issues&mdash;not because being fixed causes them, but because those issues contributed to the decision to fix them. This explanation is speculative and would require additional data to confirm.

So how does knowing this information help?

Shelters understanding which factors are associated with shorter or longer lengths of stay can be incredibly helpful in achieving their goal of finding homes for all animals. Middle-aged dogs tend to remain in shelters longer, so if shelters can better communicate the benefits of adopting a middle-aged dog, it could help improve adoption outcomes. For example, they could emphasize that middle-aged dogs often have stable personalities, are less likely to develop new health issues than older dogs, and require less training than puppies.

If the longer stays observed among fixed cats are related to behavioral issues, shelters could work on addressing those behaviors or investigate other possible explanations. Additionally, all of our data comes from an animal shelter in Austin, TX. It is possible that local cultural or personal beliefs may influence adoption patterns. Data from shelters in other regions would be useful in determining whether these patterns are consistent more broadly.
